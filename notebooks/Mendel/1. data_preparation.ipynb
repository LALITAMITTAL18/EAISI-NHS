### Load Files


```python
#1. Define paths
zip_files = [
    "../../data/external/data-pack-2016-17/Knee Replacement Provider 1617.csv.zip",
    "../../data/external/data-pack-2017-18/Knee Replacements Provider 1718.csv.zip",
    "../../data/external/data-pack-2018-19/Knee Replacement Provider 1819.csv.zip",
]
# Select specific CSVs to read from each zip file
target_csv_files = [
    "Knee Replacement Provider 1617.csv",
    "Knee Replacements Provider 1718.csv",
    "Knee Replacement Provider 1819.csv"
]

dataframes = []

for zip_path, csv_name in zip(zip_files, target_csv_files):
    with zipfile.ZipFile(zip_path, 'r') as z:
        if csv_name in z.namelist():
            with z.open(csv_name) as f:
                df = pd.read_csv(f, low_memory=False)
                dataframes.append(df)
                print(f"Successfully loaded {csv_name} from {zip_path}")
        else:
            print(f"{csv_name} not found in {zip_path}")
```

### Merge & Save File

```python
# Merge all dataframes into a single dataframe
df_prov = pd.concat(dataframes, ignore_index=True)

# Save Output
df_prov.to_csv("df_knee_prov.csv", index=False)

print("Files Merged Successfully and saved as df_prov.csv")
```

### Recode: Comorbidity Flags

These 12 columns are binary indicators (1 = condition present, 9 = not reported).  
In line with standard epidemiological practice for administrative datasets, a missing
comorbidity flag is treated as *not present* (0) rather than excluded.  
Rationale: patients with unrecorded comorbidities are unlikely to differ systematically
from patients who answered 'No'; dropping these rows would introduce selection bias.

```python
# Count patients by Comorbidities, sorted by value
for col in ["Arthritis", 
            "Cancer", 
            "Circulation", 
            "Depression", 
            "Diabetes",
            "Heart Disease", 
            "High Bp", 
            "Kidney Disease", 
            "Liver Disease",
            "Lung Disease", 
            "Nervous System", 
            "Stroke"]:
    print(f"\n{col}:")
    print(df_prov_imp[col].value_counts().sort_index())
```

```python
comorb_cols = [
    'Heart Disease', 'High Bp', 'Stroke', 'Circulation', 'Lung Disease',
    'Diabetes', 'Kidney Disease', 'Nervous System', 'Liver Disease',
    'Cancer', 'Depression', 'Arthritis'
]

# Recode 9 → 0 for all comorbidity columns
for col in comorb_cols:
    df_prov_imp[col] = df_prov_imp[col].replace(9, 0)

# Verify recode
print("Comorbidity columns after recoding 9 → 0:")
for col in comorb_cols:
    print(f"{col}: {sorted(df_prov_imp[col].unique())}")
```

```python
# Verify that all 9s have been replaced
comorb_cols = [
    'Arthritis', 'Cancer', 'Circulation', 'Depression', 'Diabetes',
    'Heart Disease', 'High Bp', 'Kidney Disease', 'Liver Disease',
    'Lung Disease', 'Nervous System', 'Stroke'
]

# Check which indicator columns exist and have null values
existing_cols = [c for c in comorb_cols if c in df_prov_imp.columns]
missing_cols  = [c for c in comorb_cols if c not in df_prov_imp.columns]

if missing_cols:
    print(f"Columns NOT found in dataframe: {missing_cols}")

null_check = df_prov_imp[existing_cols].isnull().sum()
print("Null counts for indicator columns:")
print(null_check)
```

### Standardize missing-value codes → `NaN`

The NHS PROMs data uses several conventions for missing / suppressed values:
- **`9`** — the dictionary-defined missing code for most categorical fields
- **`*`** — statistical suppression applied to small counts (< 5) by NHS Digital
- **`999`** — occasionally used for out-of-range / missing numeric entries
- **`''`** — empty string from CSV export

Converting all of these to `NaN` before any imputation or deletion step ensures that
downstream pandas methods (`.isna()`, `.fillna()`, `.dropna()`) behave consistently.

```python
df_prov_imp.replace({9: np.nan, "*": np.nan, "": np.nan, 999: np.nan}, inplace=True)
```

```python
total_missing = df_prov_imp.isna().sum().sum()
print(f'Total missing cells after standardisation: {total_missing:,}')
```

```python
# Comprehensive summary for all string variables
string_vars = ["Provider Code", "Procedure", "Year", "Age Band", "Gender"]
total_obs = len(df_prov_imp)

for var in string_vars:
    print(f"\n{'='*70}")
    print(f"{var.upper()}")
    print(f"{'='*70}")
    
    value_counts = df_prov_imp[var].value_counts(dropna=False).sort_index()
    
    summary_data = []
    for val, count in value_counts.items():
        pct = (100 * count / total_obs)
        summary_data.append({
            var: str(val) if pd.notna(val) else "Missing/NaN",
            "Count": count,
            "Percentage": round(pct, 2)
        })
    
    summary_data.append({
        var: "TOTAL",
        "Count": total_obs,
        "Percentage": 100.00
    })
    
    summary_df = pd.DataFrame(summary_data)
    print(summary_df.to_string(index=False))

```
