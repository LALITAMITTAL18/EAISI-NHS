```python
# Load original parquet file
import polars as pl
```

```python
df = pl.read_parquet(f'../Experiment/Knee/prepared_datasets/pipeline1_test.parquet')
df.shape
```

```python
freq_table = (
    df
    .group_by("Age Band")
    .agg(pl.count().alias("frequency"))
    .sort("frequency", descending=True)
)

print(freq_table)
```

```python
pl.Config.set_tbl_rows(-1)
```

```python
(
    df
    .filter(pl.col("Age Band").is_in([1, 6]))
    .group_by(["Age Band", "health_gain"])
    .len()
    .sort(["Age Band", "count"], descending=True)
)
```

```python
(
    df
    .filter(pl.col("Age Band").is_in([1, 6]))
    .group_by("Age Band")
    .agg([
        pl.col("health_gain").mean().alias("mean"),
        pl.col("health_gain").median().alias("median"),
        pl.col("health_gain").min().alias("min"),
        pl.col("health_gain").max().alias("max"),
        pl.col("health_gain").std().alias("std"),
        pl.len().alias("n")
    ])
)
```
